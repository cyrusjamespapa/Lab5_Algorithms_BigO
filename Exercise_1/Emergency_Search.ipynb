```python
LAST_NAME = "Papa"
STUDENT_ID = "TUPM-26-1927"
FAVORITE_ARTIST = "Maki"

id_digits = [int(c) for c in STUDENT_ID if c.isdigit()]
letters = [c for c in FAVORITE_ARTIST.upper() if c.isalpha()]

SEED_DIGIT = id_digits[-1]
ID_SUM = sum(id_digits)
NAME_LENGTH = len([c for c in LAST_NAME if c.isalpha()])
ARTIST_LENGTH = len(letters)
ARTIST_SUM = sum(ord(c) - ord("A") + 1 for c in letters)

print("SEED_DIGIT :", SEED_DIGIT)
print("ID_SUM :", ID_SUM)
print("NAME_LENGTH :", NAME_LENGTH)
print("ARTIST_LENGTH :", ARTIST_LENGTH)
print("ARTIST_SUM :", ARTIST_SUM)
```

```python
size = 30 + ID_SUM
start = 1000 + ID_SUM
step = NAME_LENGTH + 2

original = [start + i * step for i in range(size)]

k = (SEED_DIGIT % (len(original) - 1)) + 1     # rotation point
rotated = original[k:] + original[:k]

print("ID_SUM:", ID_SUM, "| Records:", len(original), "| first record:", original[0])
print("First", k, "records moved to the end")
print("Original:", original)
print("Rotated :", rotated)
```

```python
def linear_search(data, target):
    comparisons = 0
    for i, value in enumerate(data):
        comparisons += 1
        if value == target:
            return i, comparisons
    return -1, comparisons
```

```python
def rotated_binary_search(data, target):
    comparisons = 0
    lo, hi = 0, len(data) - 1
    while lo <= hi: 
        mid = (lo + hi) // 2
        comparisons += 1
        if data[mid] == target:
            return mid, comparisons
        comparisons += 1
        if data[lo] <= data[mid]:
            comparisons += 1
            if data[lo] <= target < data[mid]:
                hi = mid - 1
            else:
                lo = mid + 1
        else:
            comparisons += 1
            if data[mid] < target <= data[hi]:
                lo = mid + 1
            else: 
                hi = mid - 1
    return - 1, comparisons 
```

```python
target = original[(ID_SUM + ARTIST_SUM) % len(original)]
missing = target + 1          # guaranteed not in the list

print("Sanity check -> target in rotated:", target in rotated)
print()

for label, t in (("Existing incident", target), ("Missing incident", missing)):
    b_idx, b_cmp = rotated_binary_search(rotated, t)
    l_idx, l_cmp = linear_search(rotated, t)
    print(f"--- {label}: {t} ---")
    print("Rotated Binary Search:", "index", b_idx if b_idx != -1 else "NOT FOUND", "| comparisons =", b_cmp)
    print("Linear Search        :", "index", l_idx if l_idx != -1 else "NOT FOUND", "| comparisons =", l_cmp)
    print()

b_idx, b_cmp = rotated_binary_search(rotated, target)
l_idx, l_cmp = linear_search(rotated, target)
print("Comparison reduction:", l_cmp - b_cmp, "comparisons")
```

```python
print("Time complexity : Binary Search O(log n) | Linear Search O(n)")
print("Space complexity: O(1) for both (iterative, no extra list)")
print(f"Final Output: target {target} found at index {b_idx} of the rotated list.")
```
